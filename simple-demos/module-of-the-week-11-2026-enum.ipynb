```python
MONTHS = {'January':1, 'February':2, 'March':3, 'April':4, 'May':5}
MONTHS['January']

favorite_month = input("What is your favorite month? ")

if favorite_month in MONTHS:
    print(f"{favorite_month} is month number {MONTHS[favorite_month]}.")
else:
    print(f"{favorite_month} is not in the list of months.")
```

```python
from enum import IntEnum
from enum import auto

class Month(IntEnum):
    January = auto()
    February = auto()
    March = auto()
    April = auto()
    May = auto()

#Month.January.value == 1

Month.January > Month.May

#Month.January is 1

for month in Month.__members__.values():
    print(f"{month.name} is month number {month.value}.")
```

```python
from enum import Flag

class Flags(Flag):
    A = 1
    B = 2
    C = 3

print(Flags.A | Flags.C)
print(Flags.A)
```
